```python
# Programa para identificar o tamanho da camiseta

print("=== TAMANHO DA CAMISETA ===")

comprimento = float(input("Digite o comprimento da camiseta (cm): "))
largura = float(input("Digite a largura da camiseta (cm): "))

# Medidas aproximadas dos tamanhos
tamanhos = {
    "PP": (66, 50),
    "P":  (70, 54),
    "M":  (72, 56),
    "G":  (74, 58),
    "GG": (76, 60),
    "XGG": (79, 63)
}

# Verifica qual tamanho está mais próximo,
# considerando uma tolerância de até 2 cm
melhor_tamanho = None
menor_distancia = float("inf")

for tamanho, (comp, larg) in tamanhos.items():
    if abs(comprimento - comp) <= 2 and abs(largura - larg) <= 2:
        distancia = abs(comprimento - comp) + abs(largura - larg)

        if distancia < menor_distancia:
            menor_distancia = distancia
            melhor_tamanho = tamanho

if melhor_tamanho:
    print(f"\nO tamanho da camiseta é: {melhor_tamanho}")
else:
    print("\nNão foi possível identificar o tamanho da camiseta.")
    print("Verifique as medidas informadas.")
```
